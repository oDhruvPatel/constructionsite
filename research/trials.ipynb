```python
from langchain_classic.chains.llm import LLMChain
from langchain_classic.chains import SimpleSequentialChain
from langchain_classic.chains import SequentialChain
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.chains import ConversationChain

llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest"
)

mem = ConversationBufferMemory()

prompt = PromptTemplate(
    input_variables=['topic'],
    template="generate some text in 10 words text on {topic}"
)

chain = LLMChain(llm=llm, prompt=prompt, output_key="sentence", memory=mem)

prompt1 = PromptTemplate(
    input_variables=['sentence'], 
    template="summarize this text in two words: {sentence}"
)

chain1 = LLMChain(llm=llm, prompt=prompt1, output_key="summary")

main_chain = SequentialChain(
    chains=[chain, chain1],
    input_variables = ["topic"],
    output_variables = ["sentence", "summary"],
    verbose=False
)

res = main_chain.invoke({"topic":"AI"})
print(res)


```

```python
from langchain_classic.agents import initialize_agent, AgentType, load_tools

tools = load_tools(['wikipedia'], llm=llm)

agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True   # better for debugging
)

response = agent.invoke({
    "input": "When was Elon Musk born? What is his age in 2026?"
})

print(response)
```
